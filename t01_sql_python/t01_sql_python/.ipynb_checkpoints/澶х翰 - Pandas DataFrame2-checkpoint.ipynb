```python
#Vehicle Marketing Type是FH13 64T但是与前一行不同
list_new_column = ['previous row not exist']

for n in range(1,rc):
    if df['Vehicle Marketing Type'][n] == 'FH13 64T' and df['Vehicle Marketing Type'][n] != df['Vehicle Marketing Type'][n-1]:
        list_new_column.append('not same model with previous row')
    else:
        list_new_column.append('same model with previous row')

df['if_model_same_with_prev_row'] = list_new_column
df['if_model_same_with_prev_row']
```

```python
#第三种
iter_columns = zip(df['Causal Part Name'],df['Function Group 4 Digits'],df['Chassis Number'])

for x,y,z in iter_columns:
    print(str(x) +'->' + str(y) +'->' + str(z) )
```

```python
#更好阅读的方式
new_col = []

iter_columns = zip(df['Causal Part Name'],df['Function Group 4 Digits'],df['Chassis Number'])
for x,y,z in iter_columns:
    new_col.append(str(x) +'->' + str(y) +'->' + str(z))
    
df['new_col'] = new_col
df['new_col'][:20]
```

```python
#7.1 map
#算出+GST以后的total payment
def calculate_something(x):
    return float(x)*1.1
df['Total_plus_tax'] = df['Total_Payment'].map(calculate_something)
df[['Total_Payment','Total_plus_tax']][:20]
```

```python
#用lambda改写
df['Total_plus_tax'] = df['Total_Payment'].map(lambda cell: cell*1.1)
df[['Total_Payment','Total_plus_tax']][:20] 
```

```python
#将Causal Part Name第一部分拿出来
df['Causal Part first'] = df['Causal Part Name'].map(lambda x: str(x).lower().split(',')[0])
df['Causal Part first'][:10]
```

```python
#7.2 apply
```

```python
#QLD','NSW','VIC'是1.1倍，其他的1.2倍
df['Total_Payment_plus_tax'] = df.apply(lambda row: row['Total_Payment']*1.2 if row['State'] not in ['QLD', 'NSW','VIC']
                                       else row['Total_Payment']*1.1, axis = 1)
df['Total_Payment_plus_tax'] 
```

```python
#根据上面的写函数, 
def paytax(row):
    if row['State'] not in ['QLD','NSW','VIC']:
        ans = row['Total_Payment']*1.2
    else:
        ans = row['Total_Payment']*1.1    
    return ans

df['Total_Payment_plus_tax'] = df.apply(paytax,axis=1)
df['Total_Payment_plus_tax']
```

```python
#13.分析credit card数据
```

```python
#13.1 是topup的数据
df = pd.read_csv(r'student_download/topup_check.csv')
df[:5]
```

```python
df['Prev_Bal'] = df['Balance'].shift(1)
df['Prev_Bal'].fillna(0,inplace=True)
df['Prev_Crd'] = df['Card'].shift(1)
df['Prev_Crd'].fillna(0,inplace=True)
df['Topup'] = df.apply(lambda row: True if row['Balance'] > row['Prev_Bal'] and row['Trans Amt']>0 \
                       and row['Card'] == row['Prev_Crd'] else False, axis=1)
df
```

```python
#13.2 分析entropy
```

```python
import scipy.stats

def calculate_ent(ary):
    p_data = ary.value_counts()
    entrophy = scipy.stats.entropy(p_data)
    return entrophy
```

```python
df_ent = pd.DataFrame(df.groupby(['Card'])['Merchant Industry'].apply(calculate_ent)).reset_index(drop=False)
df_ent
```
